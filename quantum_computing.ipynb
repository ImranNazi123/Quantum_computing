```python
from qiskit.visualization import plot_histogram
from qiskit_aer import Aer
from qiskit import QuantumCircuit , transpile
```

```python
# x gate
from qiskit import QuantumCircuit

qc = QuantumCircuit(1, 1)
qc.x(0)
qc.measure(0, 0)

qc.draw("mpl")
```

```python
# paulie x gate , simulation 
s=Aer.get_backend('qasm_simulator')
comp_cir=transpile(qc,s)
job=s.run(comp_cir,shots=1024)
result=job.result()
counts=result.get_counts(qc)

print("counts:",counts)
plot_histogram(counts)
```

```python
# hardmand code
# initial state

from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
# initial state
# qtm circ
qc=QuantumCircuit(1)
state=Statevector.from_instruction(qc)
plot_bloch_multivector(state)
```

```python
# x gate
qc_x=QuantumCircuit(1)
qc_x.x(0)
state_x=Statevector.from_instruction(qc_x)
plot_bloch_multivector(state_x)
```

```python
# after hardmand 
qc_h=QuantumCircuit(1)
qc_h.h(0)
state_h=Statevector.from_instruction(qc_h)
plot_bloch_multivector(state_h)
```

```python
# z gate
qc_hz=QuantumCircuit(1)
qc_hz.h(0)
qc_hz.z(0)

state_hz=Statevector.from_instruction(qc_hz)
plot_bloch_multivector(state_hz)
```
